```python
import pandas as pd
import numpy as np
import io
import re

# File/sheet configuration
source_config = [
    ("7Валовый охват рус.xlsx", "Всего",
     "Higher", "gross_enrollment", "Total", 0),

    ("7Валовый охват рус.xlsx", "мужчины",
     "Higher", "gross_enrollment", "Male", 0),

    ("7Валовый охват рус.xlsx", "женщины",
     "Higher", "gross_enrollment", "Female", 0),

    ("5.1. Число организаций высшего образования.xls",
     "показатель - число Вузов",
     "Higher", "institutions", "Total", 1),

    ("3.2. Выпуск студентов организаций высшего образования.xlsx",
     "Показатель Выпуск всего",
     "Higher", "graduates", "Total", 1),

    ("3.2. Выпуск студентов организаций высшего образования.xlsx",
     "показатель выпуск - женщины",
     "Higher", "graduates", "Female", 1),

    ("5.4. Прием студентов организаций высшего образования.xls",
     "Прием всего",
     "Higher", "admissions", "Total", 1),

    ("5.4. Прием студентов организаций высшего образования.xls",
     "Женщины",
     "Higher", "admissions", "Female", 1),

    ("5.2. Численность студентов организаций высшего образования.xls",
     "Численность студентов",
     "Higher", "enrollment", "Total", 1),

    ("5.2. Численность студентов организаций высшего образования.xls",
     "Женщины",
     "Higher", "enrollment", "Female", 1),

    ("прием ТиПО рус(2).xls", "Показатель",
     "TVET", "admissions", "Total", 1),

    ("3.4. Численность учащихся в общеобразовательных школах.xlsx",
     "Всего", "School", "enrollment", "Total", 1),

    ("3.4. Численность учащихся в общеобразовательных школах.xlsx",
     "Девочки", "School", "enrollment", "Female", 1),

    ("1.7. Численность детей в дошкольных орг.xlsx",
     "Числ.детей в дошк.орг.",
     "Preschool", "enrollment", "Total", 1),

    ("1.7. Численность детей в дошкольных орг.xlsx",
     "девочки", "Preschool", "enrollment", "Female", 1),
]

def parse_year(value):
    """Recognize calendar years and academic years."""
    if pd.isna(value):
        return None, None

    s = str(value).strip()
    match = re.fullmatch(r"(20\d{2})(?:/(\d{2,4}))?", s)

    if not match:
        return None, None

    year = int(match.group(1))

    if not 2000 <= year <= 2025:
        return None, None

    if match.group(2):
        end = int(match.group(2)[-2:])
        if end != (year + 1) % 100:
            return None, None
        return year, "academic"

    return year, "calendar"


def parse_workbook_table(
    content, sheet, level, indicator, gender, region_col
):
    df = pd.read_excel(
        io.BytesIO(content),
        sheet_name=sheet,
        header=None,
        dtype=object
    )

    # Detect the row containing the most years
    best_row = None
    best_cols = []

    for row in range(min(15, len(df))):
        year_cols = []

        for col in range(df.shape[1]):
            year, period = parse_year(df.iat[row, col])

            if year is not None:
                year_cols.append((col, year, period))

        if len(year_cols) > len(best_cols):
            best_row = row
            best_cols = year_cols

    if len(best_cols) < 5:
        raise ValueError(
            f"No valid year header found in {sheet}"
        )

    records = []

    for row in range(best_row + 1, len(df)):
        region = df.iat[row, region_col]

        if pd.isna(region):
            continue

        region = str(region).strip()

        # Main regional tables have national + regional rows.
        # Stop at notes, footers and additional sections.
        if region.lower().startswith(
            ("примечание", "источник", "сноска")
        ):
            break

        for col, year, period in best_cols:
            raw = df.iat[row, col]

            raw_str = "" if pd.isna(raw) else str(raw).strip()
            raw_str = raw_str.replace("\u00a0", " ")

            value = pd.to_numeric(
                raw_str.replace(" ", "").replace(",", "."),
                errors="coerce"
            )

            records.append({
                "year": year,
                "period_type": period,
                "region": region,
                "education_level": level,
                "indicator": indicator,
                "gender": gender,
                "value": value,
                "raw_value": raw_str,
                "source_sheet": sheet
            })

    return pd.DataFrame(records)


all_tables = []
errors = []

for filename, sheet, level, indicator, gender, region_col in source_config:
    try:
        if filename not in uploaded:
            raise FileNotFoundError(filename)

        result = parse_workbook_table(
            uploaded[filename],
            sheet, level, indicator, gender, region_col
        )

        result["source_file"] = filename
        all_tables.append(result)

        print(f"Loaded: {filename} | {sheet}: {len(result)} rows")

    except Exception as error:
        errors.append((filename, sheet, str(error)))
        print(f"ERROR: {filename} | {sheet} -> {error}")

if not all_tables:
    raise RuntimeError("No source tables were parsed.")

education_long = pd.concat(
    all_tables, ignore_index=True
)

display(education_long.head(15))

print("Total extracted rows:", len(education_long))
print("Files/sheets with errors:", len(errors))
display(pd.DataFrame(errors, columns=["file", "sheet", "error"]))
```

```python
import pandas as pd
import numpy as np
import re

# Keep the original extraction unchanged
education_clean = education_long.copy()

# 1. Standardize regional names
region_aliases = {
    "Республика Казахстан": "Kazakhstan",
    "Абай": "Abay",
    "Акмолинская": "Akmola",
    "Актюбинская": "Aktobe",
    "Алматинская": "Almaty Region",
    "Атырауская": "Atyrau",
    "Восточно-Казахстанская": "East Kazakhstan",
    "Жамбылская": "Zhambyl",
    "Жетису": "Zhetysu",
    "Жетісу": "Zhetysu",
    "Западно-Казахстанская": "West Kazakhstan",
    "Карагандинская": "Karaganda",
    "Костанайская": "Kostanay",
    "Кызылординская": "Kyzylorda",
    "Мангистауская": "Mangystau",
    "Павлодарская": "Pavlodar",
    "Северо-Казахстанская": "North Kazakhstan",
    "Туркестанская": "Turkestan",
    "Улытау": "Ulytau",
    "Ұлытау": "Ulytau",
    "г.Алматы": "Almaty City",
    "город Алматы": "Almaty City",
    "г.Астана": "Astana",
    "город Астана": "Astana",
    "г.Шымкент": "Shymkent",
    "город Шымкент": "Shymkent",
}

def normalize_region(name):
    if pd.isna(name):
        return None

    name = str(name).strip()

    # Remove footnote symbols attached to region names
    name = re.sub(r"\*+$", "", name).strip()

    return region_aliases.get(name)


education_clean["region_standard"] = (
    education_clean["region"].apply(normalize_region)
)

invalid_regions = education_clean[
    education_clean["region_standard"].isna()
]

print("Rows with invalid region labels:",
      len(invalid_regions))

education_clean = education_clean[
    education_clean["region_standard"].notna()
].copy()


# 2. Distinguish school categories
education_clean["school_type"] = "Not applicable"

school_mask = education_clean["education_level"] == "School"

# The workbook has two consecutive regional blocks:
# daytime schools first, evening schools second.
# Preserve both blocks as separate indicators.
school_rows = education_clean.loc[school_mask].copy()

school_keys = [
    "source_file", "source_sheet",
    "year", "region_standard"
]

school_rows["section_order"] = (
    school_rows.groupby(school_keys, sort=False)
    .cumcount()
)

section_counts = school_rows.groupby(
    school_keys
).size()

if section_counts.max() > 2:
    raise ValueError(
        "Unexpected third school section detected. "
        "Inspect workbook before continuing."
    )

school_rows["school_type"] = np.where(
    school_rows["section_order"] == 0,
    "Daytime",
    "Evening"
)

education_clean.loc[
    school_rows.index, "school_type"
] = school_rows["school_type"]

# 3. Preserve official missing-value meaning
def classify_value(raw, value):
    if pd.notna(value):
        return "Observed"

    raw = str(raw).strip()

    if raw == "-":
        return "Phenomenon absent"
    elif raw in ["...", "…"]:
        return "Not available"
    elif raw.lower() in ["x", "х"]:
        return "Confidential"
    elif raw == "":
        return "Blank"
    else:
        return "Other missing"


education_clean["value_status"] = (
    education_clean.apply(
        lambda row: classify_value(
            row["raw_value"], row["value"]
        ),
        axis=1
    )
)

# 4. Check uniqueness
keys = [
    "year",
    "period_type",
    "region_standard",
    "education_level",
    "indicator",
    "gender",
    "school_type"
]

duplicates = education_clean.duplicated(
    subset=keys, keep=False
)

print("Remaining duplicate-key rows:",
      duplicates.sum())

if duplicates.any():
    display(
        education_clean.loc[duplicates]
        .sort_values(keys)
        .head(30)
    )

    raise ValueError(
        "Duplicate indicator keys remain. "
        "Resolve these before analysis."
    )

# 5. Create analysis-ready columns
education_clean["region"] = (
    education_clean["region_standard"]
)

education_clean["year"] = (
    education_clean["year"].astype(int)
)

education_clean["value"] = pd.to_numeric(
    education_clean["value"], errors="coerce"
)

# Preserve the units
education_clean["unit"] = np.where(
    education_clean["indicator"] == "gross_enrollment",
    "Percent",
    np.where(
        education_clean["indicator"] == "institutions",
        "Institutions",
        "Persons"
    )
)

# 6. Final summary
print("\nCleaned dataset shape:",
      education_clean.shape)

print("\nEducation levels:")
print(
    education_clean["education_level"]
    .value_counts()
)

print("\nValue statuses:")
print(
    education_clean["value_status"]
    .value_counts()
)

print("\nUnique geographic units:")
print(
    education_clean["region"].nunique()
)

display(education_clean.head(10))
```

```python
# STEP 7 — DATA INTEGRITY AUDIT

import pandas as pd
import numpy as np

df = education_clean.copy()

# 1. Verify geographic coverage
all_regions = set(
    df.loc[df["region"] != "Kazakhstan", "region"]
)

coverage = []

for level in sorted(df["education_level"].unique()):
    level_regions = set(
        df.loc[
            (df["education_level"] == level) &
            (df["region"] != "Kazakhstan"),
            "region"
        ]
    )

    coverage.append({
        "education_level": level,
        "regions_present": len(level_regions),
        "missing_regions": ", ".join(
            sorted(all_regions - level_regions)
        ) or "None"
    })

print("GEOGRAPHIC COVERAGE")
display(pd.DataFrame(coverage))


# 2. Missing observations by year and level
missing_by_year = (
    df.assign(missing=df["value"].isna())
      .groupby(["education_level", "year"], as_index=False)
      .agg(
          records=("missing", "size"),
          missing=("missing", "sum")
      )
)

missing_by_year["missing_pct"] = (
    100 * missing_by_year["missing"] /
    missing_by_year["records"]
).round(2)

print("\nMISSINGNESS BY YEAR")
display(missing_by_year.tail(20))


# 3. Validate Female <= Total
comparison_keys = [
    "year", "period_type", "region",
    "education_level", "indicator", "school_type"
]

gender_check = df.pivot_table(
    index=comparison_keys,
    columns="gender",
    values="value",
    aggfunc="first"
)

if {"Female", "Total"}.issubset(gender_check.columns):
    valid_pairs = gender_check.dropna(
        subset=["Female", "Total"]
    ).copy()

    invalid_gender = valid_pairs[
        (valid_pairs["Female"] < 0) |
        (valid_pairs["Female"] > valid_pairs["Total"])
    ]

    print("\nFEMALE > TOTAL VIOLATIONS:",
          len(invalid_gender))

    display(invalid_gender.head(20))


# 4. Validate national totals against regional sums
# Only compare complete, additive count indicators
count_indicators = [
    "enrollment", "admissions",
    "graduates", "institutions"
]

regional_df = df[
    (df["region"] != "Kazakhstan") &
    (df["indicator"].isin(count_indicators))
]

national_df = df[
    (df["region"] == "Kazakhstan") &
    (df["indicator"].isin(count_indicators))
]

group_cols = [
    "year", "period_type",
    "education_level", "indicator",
    "gender", "school_type"
]

regional_sum = regional_df.groupby(
    group_cols, dropna=False
).agg(
    regional_total=("value",
                    lambda s: s.sum(min_count=1)),
    valid_regions=("value", "count"),
    listed_regions=("region", "nunique")
).reset_index()

national = national_df[
    group_cols + ["value"]
].rename(columns={"value": "national_total"})

reconciliation = national.merge(
    regional_sum,
    on=group_cols,
    how="left"
)

reconciliation["difference"] = (
    reconciliation["national_total"] -
    reconciliation["regional_total"]
)

# Restrict comparisons to complete coverage
reconciliation["complete"] = (
    reconciliation["valid_regions"] ==
    reconciliation["listed_regions"]
)

print("\nNATIONAL VS REGIONAL TOTALS")
display(
    reconciliation[
        reconciliation["complete"] &
        reconciliation["national_total"].notna() &
        reconciliation["difference"].abs().gt(0)
    ].head(30)
)

print("\nIncomplete groups:",
      (~reconciliation["complete"]).sum())

# 5. Final summary
print("\nDATASET SUMMARY")
print("Rows:", len(df))
print("Observed:", df["value"].notna().sum())
print("Missing:", df["value"].isna().sum())
print("Duplicate keys:",
      df.duplicated(subset=[
          "year", "period_type", "region",
          "education_level", "indicator",
          "gender", "school_type"
      ]).sum())
```

```python
# STEP 11C — CSV, PARQUET AND MEMORY OPTIMIZATION

from pathlib import Path
import pandas as pd
import numpy as np
import time
import sys
import subprocess

# Install Parquet support
subprocess.check_call([
    sys.executable, "-m", "pip", "-q", "install", "pyarrow"
])

PROJECT = Path("/content/kazakhstan_education_project")
DATA = PROJECT / "data"
DOCS = PROJECT / "docs"
DOCS.mkdir(exist_ok=True)

csv_path = DATA / "education_clean.csv"
parquet_path = DATA / "education_optimized.parquet"

# 1. Original dataset
original = pd.read_csv(csv_path)

# 2. Optimize low-cardinality text columns
category_cols = [
    "period_type", "region", "education_level",
    "indicator", "gender", "source_sheet",
    "source_file", "region_standard",
    "school_type", "value_status", "unit",
    "quality_flag"
]

dtype_map = {
    col: "category" for col in category_cols
}
dtype_map["year"] = "int16"

optimized = original.astype(dtype_map)

# 3. Verify the optimization preserves the data
restored = optimized.astype({
    col: "object" for col in category_cols
})

pd.testing.assert_frame_equal(
    original, restored, check_dtype=False
)

# 4. Save optimized dataset as Parquet
optimized.to_parquet(parquet_path, index=False)

# 5. Measure loading time, DataFrame memory and disk size
def measure_loading(name, loader, path, runs=5):
    loader()  # Warm-up
    times = []

    for _ in range(runs):
        start = time.perf_counter()
        result = loader()
        times.append(
            (time.perf_counter() - start) * 1000
        )

    assert len(result) == len(original)

    return {
        "Method": name,
        "Read time (ms)": round(np.median(times), 2),
        "Memory (MB)": round(
            result.memory_usage(deep=True).sum() / 1024**2,
            3
        ),
        "File size (MB)": round(
            path.stat().st_size / 1024**2, 3
        )
    }


results = [
    measure_loading(
        "Standard CSV",
        lambda: pd.read_csv(csv_path),
        csv_path
    ),
    measure_loading(
        "Optimized CSV",
        lambda: pd.read_csv(csv_path, dtype=dtype_map),
        csv_path
    ),
    measure_loading(
        "Optimized Parquet",
        lambda: pd.read_parquet(parquet_path),
        parquet_path
    )
]

benchmark_df = pd.DataFrame(results)

# Verify the optimized formats load the same observations
for loaded in [
    pd.read_csv(csv_path, dtype=dtype_map),
    pd.read_parquet(parquet_path)
]:
    recovered = loaded.astype({
        col: "object" for col in category_cols
    })
    pd.testing.assert_frame_equal(
        original, recovered, check_dtype=False
    )

# Save results for the report and final ZIP
benchmark_df.to_csv(
    DOCS / "benchmark_results.csv",
    index=False
)

display(benchmark_df)

baseline_memory = benchmark_df.loc[0, "Memory (MB)"]
optimized_memory = benchmark_df.loc[2, "Memory (MB)"]

saving = 100 * (
    1 - optimized_memory / baseline_memory
)

print(f"\nDataFrame memory reduction: {saving:.1f}%")
print("All data-equivalence checks passed.")
print("Benchmark saved to docs/benchmark_results.csv")
print("Optimized Parquet saved to data/education_optimized.parquet")
```

```python
# STEP 12C — IQR OUTLIER ANALYSIS

eda_2025 = df[
    (df["region"] != "Kazakhstan") &
    (df["year"] == 2025) &
    (df["gender"] == "Total") &
    (df["school_type"] != "Evening") &
    (df["analysis_value"].notna())
].copy()

groups = [
    "education_level",
    "indicator",
    "unit"
]

outlier_results = []
flagged_records = []

for key, group in eda_2025.groupby(groups):

    if len(group) < 4:
        continue

    q1 = group["analysis_value"].quantile(0.25)
    q3 = group["analysis_value"].quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = group[
        (group["analysis_value"] < lower) |
        (group["analysis_value"] > upper)
    ].copy()

    outliers["lower_bound"] = lower
    outliers["upper_bound"] = upper

    flagged_records.append(outliers)

    outlier_results.append({
        "education_level": key[0],
        "indicator": key[1],
        "unit": key[2],
        "observations": len(group),
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "lower_bound": lower,
        "upper_bound": upper,
        "outlier_count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_results)

print("IQR OUTLIER SUMMARY")
display(outlier_summary.round(2))

if flagged_records:
    outlier_details = pd.concat(
        flagged_records,
        ignore_index=True
    )

    print("\nFLAGGED REGIONAL OBSERVATIONS")
    display(
        outlier_details[
            [
                "region", "education_level",
                "indicator", "analysis_value",
                "lower_bound", "upper_bound"
            ]
        ].sort_values(
            ["education_level", "indicator"]
        )
    )

    outlier_details.to_csv(
        "eda_outliers.csv", index=False
    )

outlier_summary.to_csv(
    "eda_outlier_summary.csv",
    index=False
)
```

```python
# STEP 15A — CHI-SQUARE TEST BY EDUCATION LEVEL

import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

chi_results = []

for level in ["Preschool", "School", "Higher"]:

    subset = regional_gender[
        regional_gender["education_level"] == level
    ].copy()

    contingency = subset[
        ["Female", "Male"]
    ].to_numpy()

    chi2, p, dof, expected = chi2_contingency(
        contingency
    )

    n = contingency.sum()

    cramers_v = np.sqrt(
        chi2 / (
            n * min(
                contingency.shape[0] - 1,
                contingency.shape[1] - 1
            )
        )
    )

    chi_results.append({
        "education_level": level,
        "chi_square": chi2,
        "degrees_of_freedom": dof,
        "p_value": p,
        "cramers_v": cramers_v,
        "minimum_expected_count": expected.min()
    })

chi_results = pd.DataFrame(chi_results)

print("CHI-SQUARE TEST RESULTS")
display(chi_results.round(6))

chi_results.to_csv(
    "chi_square_results.csv",
    index=False
)
```

```python
# STEP 15B — ONE-SAMPLE T-TEST

from scipy.stats import ttest_1samp, t

t_results = []

for level in ["Preschool", "School", "Higher"]:

    shares = regional_gender.loc[
        regional_gender["education_level"] == level,
        "female_share_pct"
    ].dropna()

    result = ttest_1samp(
        shares, popmean=50
    )

    n = len(shares)
    mean = shares.mean()
    sd = shares.std(ddof=1)
    se = sd / np.sqrt(n)

    margin = t.ppf(0.975, n - 1) * se

    t_results.append({
        "education_level": level,
        "regions": n,
        "mean_female_share": mean,
        "std_dev": sd,
        "t_statistic": result.statistic,
        "p_value": result.pvalue,
        "CI_lower": mean - margin,
        "CI_upper": mean + margin
    })

t_results = pd.DataFrame(t_results)

print("ONE-SAMPLE T-TEST RESULTS")
display(t_results.round(5))

t_results.to_csv(
    "one_sample_ttest_results.csv",
    index=False
)
```

```python
# STEP 15C — PAIRED T-TEST:
# SCHOOL VS HIGHER EDUCATION

from scipy.stats import ttest_rel, t

comparison = regional_gender.pivot(
    index="region",
    columns="education_level",
    values="female_share_pct"
)

paired = comparison[
    ["School", "Higher"]
].dropna().copy()

paired["difference_pp"] = (
    paired["Higher"] - paired["School"]
)

test = ttest_rel(
    paired["Higher"],
    paired["School"]
)

n = len(paired)
mean_diff = paired["difference_pp"].mean()
sd_diff = paired["difference_pp"].std(ddof=1)

margin = (
    t.ppf(0.975, n - 1) *
    sd_diff / np.sqrt(n)
)

print("PAIRED COMPARISON: HIGHER - SCHOOL")

print("Regions:", n)
print("Mean difference:", round(mean_diff, 3),
      "percentage points")
print("t-statistic:", round(test.statistic, 4))
print("p-value:", test.pvalue)

print(
    "95% confidence interval:",
    (round(mean_diff - margin, 3),
     round(mean_diff + margin, 3))
)

print("\nREGIONAL DIFFERENCES")
display(
    paired.sort_values(
        "difference_pp",
        ascending=False
    ).round(2)
)

paired.to_csv(
    "paired_gender_comparison.csv"
)
```

```python
# STEP 16 — ROBUSTNESS CHECKS

import numpy as np
import pandas as pd

from scipy.stats import (
    ttest_1samp,
    wilcoxon,
    shapiro
)
from statsmodels.stats.multitest import multipletests

# Previously calculated paired comparison
differences = paired["difference_pp"].dropna()

# 1. Normality diagnostic
shapiro_result = shapiro(differences)

print("NORMALITY DIAGNOSTIC")
print("Shapiro-Wilk p-value:",
      round(shapiro_result.pvalue, 5))

# 2. Nonparametric sensitivity check
wilcoxon_result = wilcoxon(differences)

print("\nWILCOXON SIGNED-RANK TEST")
print("Statistic:", wilcoxon_result.statistic)
print("p-value:", wilcoxon_result.pvalue)

# 3. Leave-one-region-out sensitivity analysis
sensitivity = []

for region in differences.index:
    remaining = differences.drop(region)

    result = ttest_1samp(remaining, 0)

    sensitivity.append({
        "excluded_region": region,
        "mean_difference_pp": remaining.mean(),
        "p_value": result.pvalue
    })

sensitivity_df = pd.DataFrame(sensitivity)

print("\nLEAVE-ONE-REGION-OUT ANALYSIS")
display(
    sensitivity_df.sort_values("p_value").round(5)
)

print("Significant results (p < 0.05):",
      (sensitivity_df["p_value"] < 0.05).sum(),
      "out of", len(sensitivity_df))

# 4. Multiple testing correction
# Treat all seven previous tests as one exploratory family
p_values = (
    chi_results["p_value"].tolist() +
    t_results["p_value"].tolist() +
    [test.pvalue]
)

test_names = (
    ["Chi-square: " + x
     for x in chi_results["education_level"]] +
    ["One-sample t-test: " + x
     for x in t_results["education_level"]] +
    ["Paired t-test: Higher vs School"]
)

reject, adjusted_p, _, _ = multipletests(
    p_values,
    alpha=0.05,
    method="holm"
)

multiple_tests = pd.DataFrame({
    "test": test_names,
    "original_p": p_values,
    "holm_adjusted_p": adjusted_p,
    "reject_at_5pct": reject
})

print("\nMULTIPLE-TESTING CORRECTION")
display(multiple_tests.round(6))

# Save results
sensitivity_df.to_csv(
    "gender_sensitivity_analysis.csv",
    index=False
)

multiple_tests.to_csv(
    "gender_multiple_testing.csv",
    index=False
)
```

```python
# STEP 21A — PREPARE THE FORECASTING DATASET

import pandas as pd
import numpy as np

admissions = national_ts[
    (national_ts["education_level"] == "Higher") &
    (national_ts["indicator"] == "admissions")
].sort_values("year")[["year", "value"]].copy()

admissions = admissions.set_index("year")["value"]

assert admissions.index.is_unique
assert admissions.notna().all()
assert admissions.index.tolist() == list(range(2000, 2026))

test_years = list(range(2021, 2026))

print("Forecast evaluation years:", test_years)
print("Historical observations:", len(admissions))

display(
    admissions.loc[2018:2025].to_frame(
        "Actual admissions"
    )
)
```

```python
# STEP 21B — FORECASTING FUNCTIONS

def naive_forecast(history):
    return float(history.iloc[-1])


def moving_average_forecast(history, window):
    if len(history) < window:
        raise ValueError("Insufficient history.")

    return float(history.iloc[-window:].mean())


def linear_trend_forecast(history, window=5):
    if len(history) < window:
        raise ValueError("Insufficient history.")

    y = history.iloc[-window:].to_numpy(dtype=float)
    x = np.arange(window, dtype=float)

    slope, intercept = np.polyfit(x, y, 1)

    # Predict one year after the fitted window
    return float(slope * window + intercept)


methods = {
    "Naive": naive_forecast,
    "MA (3 years)": lambda h: moving_average_forecast(h, 3),
    "MA (5 years)": lambda h: moving_average_forecast(h, 5),
    "Linear regression (5 years)": (
        lambda h: linear_trend_forecast(h, 5)
    )
}

# Test all four methods using 2021
history = admissions.loc[:2020]

for name, method in methods.items():
    print(name, ":", round(method(history), 2))
```

```python
# STEP 21C — CHRONOLOGICAL FORECAST EVALUATION

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)

forecast_records = []

for year in test_years:
    history = admissions.loc[admissions.index < year]
    actual = admissions.loc[year]

    for name, method in methods.items():
        prediction = method(history)

        forecast_records.append({
            "year": year,
            "method": name,
            "actual": actual,
            "predicted": prediction,
            "absolute_error": abs(actual - prediction)
        })

forecast_df = pd.DataFrame(forecast_records)

metrics = []

for name, group in forecast_df.groupby("method"):
    actual = group["actual"].to_numpy()
    predicted = group["predicted"].to_numpy()

    metrics.append({
        "method": name,
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(
            mean_squared_error(actual, predicted)
        ),
        "MAPE_pct": (
            mean_absolute_percentage_error(
                actual, predicted
            ) * 100
        )
    })

metrics_df = pd.DataFrame(metrics).sort_values("MAE")

print("FORECAST ACCURACY COMPARISON")
display(metrics_df.round(2))

print("\nYEAR-BY-YEAR FORECASTS")
display(
    forecast_df.pivot(
        index="year",
        columns="method",
        values="predicted"
    ).round(2)
)

forecast_df.to_csv(
    "forecast_predictions.csv",
    index=False
)

metrics_df.to_csv(
    "forecast_accuracy.csv",
    index=False
)
```

```python
# STEP 21D — EXECUTION TIME AND MEMORY COMPARISON

import time
import tracemalloc

benchmark_results = []

repetitions = 200

for name, method in methods.items():

    histories = [
        admissions.loc[admissions.index < year]
        for year in test_years
    ]

    # Warm-up
    for history in histories:
        method(history)

    tracemalloc.start()
    start = time.perf_counter()

    for _ in range(repetitions):
        for history in histories:
            method(history)

    elapsed = time.perf_counter() - start
    _, peak_memory = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    n_predictions = repetitions * len(histories)

    benchmark_results.append({
        "method": name,
        "time_per_prediction_ms": (
            elapsed / n_predictions * 1000
        ),
        "peak_python_memory_kb": peak_memory / 1024
    })

benchmark_df = pd.DataFrame(benchmark_results)

final_comparison = metrics_df.merge(
    benchmark_df,
    on="method"
).sort_values("MAE")

print("COMPLETE METHOD COMPARISON")
display(final_comparison.round(4))

final_comparison.to_csv(
    "forecast_method_comparison.csv",
    index=False
)
```

```python
# STEP 22D.1 — DEFINE TEN ANALYTICAL SQL QUERIES

sql_queries = {}

# Q1 — National education indicators for 2025
# SQL concepts: JOIN, filtering, ordering

sql_queries["Q1_National_Indicators"] = """
SELECT
    l.level_name AS education_level,
    i.indicator_name AS indicator,
    s.school_type_name AS school_type,
    o.value
FROM observations o
JOIN regions r ON o.region_id = r.region_id
JOIN education_levels l ON o.level_id = l.level_id
JOIN indicators i ON o.indicator_id = i.indicator_id
JOIN genders g ON o.gender_id = g.gender_id
JOIN school_types s
    ON o.school_type_id = s.school_type_id
WHERE o.year = 2025
  AND r.region_name = 'Kazakhstan'
  AND g.gender_name = 'Total'
  AND s.school_type_name != 'Evening'
ORDER BY l.level_name, i.indicator_name;
"""


# Q2 — Top regions by school enrollment
# SQL concepts: JOIN, ORDER BY, LIMIT

sql_queries["Q2_School_Enrollment"] = """
SELECT
    r.region_name AS region,
    CAST(o.value AS INTEGER) AS students
FROM observations o
JOIN regions r ON o.region_id = r.region_id
JOIN education_levels l ON o.level_id = l.level_id
JOIN indicators i ON o.indicator_id = i.indicator_id
JOIN genders g ON o.gender_id = g.gender_id
JOIN school_types s
    ON o.school_type_id = s.school_type_id
WHERE o.year = 2025
  AND l.level_name = 'School'
  AND i.indicator_name = 'enrollment'
  AND g.gender_name = 'Total'
  AND s.school_type_name = 'Daytime'
  AND r.region_name != 'Kazakhstan'
ORDER BY students DESC
LIMIT 10;
"""


# Q3 — Female shares in higher education
# SQL concepts: CTE, conditional aggregation, HAVING

sql_queries["Q3_Higher_Gender_Shares"] = """
WITH gender_counts AS (
    SELECT
        r.region_name AS region,
        MAX(CASE WHEN g.gender_name = 'Female'
            THEN o.value END) AS female,
        MAX(CASE WHEN g.gender_name = 'Total'
            THEN o.value END) AS total
    FROM observations o
    JOIN regions r ON o.region_id = r.region_id
    JOIN education_levels l ON o.level_id = l.level_id
    JOIN indicators i ON o.indicator_id = i.indicator_id
    JOIN genders g ON o.gender_id = g.gender_id
    WHERE o.year = 2025
      AND l.level_name = 'Higher'
      AND i.indicator_name = 'enrollment'
      AND r.region_name != 'Kazakhstan'
    GROUP BY r.region_name
)
SELECT
    region,
    female,
    total,
    ROUND(100.0 * female / total, 2) AS female_pct
FROM gender_counts
WHERE female IS NOT NULL
  AND total > 0
  AND female BETWEEN 0 AND total
ORDER BY female_pct DESC;
"""


# Q4 — Concentration in three major cities
# SQL concepts: CTE, GROUP BY, conditional aggregation

sql_queries["Q4_Major_City_Concentration"] = """
WITH education_counts AS (
    SELECT
        r.region_name AS region,
        i.indicator_name AS indicator,
        o.value
    FROM observations o
    JOIN regions r ON o.region_id = r.region_id
    JOIN education_levels l ON o.level_id = l.level_id
    JOIN indicators i ON o.indicator_id = i.indicator_id
    JOIN genders g ON o.gender_id = g.gender_id
    WHERE o.year = 2025
      AND l.level_name = 'Higher'
      AND g.gender_name = 'Total'
      AND i.indicator_name IN
          ('institutions', 'enrollment', 'graduates')
)
SELECT
    indicator,
    SUM(CASE WHEN region IN
        ('Almaty City', 'Astana', 'Shymkent')
        THEN value ELSE 0 END) AS three_city_total,
    MAX(CASE WHEN region = 'Kazakhstan'
        THEN value END) AS national_total,
    ROUND(
        100.0 * SUM(CASE WHEN region IN
            ('Almaty City', 'Astana', 'Shymkent')
            THEN value ELSE 0 END)
        / MAX(CASE WHEN region = 'Kazakhstan'
            THEN value END),
        2
    ) AS combined_share_pct
FROM education_counts
GROUP BY indicator
ORDER BY indicator;
"""


# Q5 — Regional university enrollment ranking
# SQL concepts: window functions RANK and SUM OVER

sql_queries["Q5_Enrollment_Ranking"] = """
WITH regional_enrollment AS (
    SELECT
        r.region_name AS region,
        o.value AS students
    FROM observations o
    JOIN regions r ON o.region_id = r.region_id
    JOIN education_levels l ON o.level_id = l.level_id
    JOIN indicators i ON o.indicator_id = i.indicator_id
    JOIN genders g ON o.gender_id = g.gender_id
    WHERE o.year = 2025
      AND l.level_name = 'Higher'
      AND i.indicator_name = 'enrollment'
      AND g.gender_name = 'Total'
      AND r.region_name != 'Kazakhstan'
)
SELECT
    region,
    students,
    RANK() OVER (
        ORDER BY students DESC
    ) AS enrollment_rank,
    ROUND(
        100.0 * students / SUM(students) OVER (),
        2
    ) AS national_share_pct
FROM regional_enrollment
ORDER BY enrollment_rank;
"""


# Q6 — Historical higher education admissions growth
# SQL concepts: LAG window function, CTE

sql_queries["Q6_Admissions_Growth"] = """
WITH annual_admissions AS (
    SELECT
        o.year,
        o.value AS admissions,
        LAG(o.value) OVER (
            ORDER BY o.year
        ) AS previous_year
    FROM observations o
    JOIN regions r ON o.region_id = r.region_id
    JOIN education_levels l ON o.level_id = l.level_id
    JOIN indicators i ON o.indicator_id = i.indicator_id
    JOIN genders g ON o.gender_id = g.gender_id
    WHERE r.region_name = 'Kazakhstan'
      AND l.level_name = 'Higher'
      AND i.indicator_name = 'admissions'
      AND g.gender_name = 'Total'
)
SELECT
    year,
    admissions,
    previous_year,
    ROUND(
        100.0 * (admissions - previous_year)
        / NULLIF(previous_year, 0),
        2
    ) AS growth_pct
FROM annual_admissions
ORDER BY year;
"""


# Q7 — Regions with above-average higher enrollment
# SQL concepts: subquery, AVG

sql_queries["Q7_Above_Average_Regions"] = """
SELECT
    r.region_name AS region,
    o.value AS enrollment
FROM observations o
JOIN regions r ON o.region_id = r.region_id
JOIN education_levels l ON o.level_id = l.level_id
JOIN indicators i ON o.indicator_id = i.indicator_id
JOIN genders g ON o.gender_id = g.gender_id
WHERE o.year = 2025
  AND l.level_name = 'Higher'
  AND i.indicator_name = 'enrollment'
  AND g.gender_name = 'Total'
  AND r.region_name != 'Kazakhstan'
  AND o.value > (
      SELECT AVG(o2.value)
      FROM observations o2
      JOIN regions r2 ON o2.region_id = r2.region_id
      JOIN education_levels l2 ON o2.level_id = l2.level_id
      JOIN indicators i2 ON o2.indicator_id = i2.indicator_id
      JOIN genders g2 ON o2.gender_id = g2.gender_id
      WHERE o2.year = 2025
        AND l2.level_name = 'Higher'
        AND i2.indicator_name = 'enrollment'
        AND g2.gender_name = 'Total'
        AND r2.region_name != 'Kazakhstan'
  )
ORDER BY enrollment DESC;
"""


# Q8 — Students per higher education institution
# SQL concepts: CTE, conditional aggregation

sql_queries["Q8_Institutional_Scale"] = """
WITH regional_counts AS (
    SELECT
        r.region_name AS region,
        MAX(CASE WHEN i.indicator_name = 'enrollment'
            THEN o.value END) AS students,
        MAX(CASE WHEN i.indicator_name = 'institutions'
            THEN o.value END) AS institutions
    FROM observations o
    JOIN regions r ON o.region_id = r.region_id
    JOIN education_levels l ON o.level_id = l.level_id
    JOIN indicators i ON o.indicator_id = i.indicator_id
    JOIN genders g ON o.gender_id = g.gender_id
    WHERE o.year = 2025
      AND l.level_name = 'Higher'
      AND g.gender_name = 'Total'
      AND i.indicator_name IN
          ('enrollment', 'institutions')
      AND r.region_name != 'Kazakhstan'
    GROUP BY r.region_name
)
SELECT
    region,
    students,
    institutions,
    ROUND(
        students / NULLIF(institutions, 0),
        2
    ) AS students_per_institution
FROM regional_counts
WHERE students IS NOT NULL
  AND institutions > 0
ORDER BY students_per_institution DESC;
"""


# Q9 — Compare higher education and TVET admissions
# SQL concepts: GROUP BY, HAVING, conditional aggregation

sql_queries["Q9_TVET_vs_Higher_Admissions"] = """
SELECT
    r.region_name AS region,
    MAX(CASE WHEN l.level_name = 'Higher'
        THEN o.value END) AS higher_admissions,
    MAX(CASE WHEN l.level_name = 'TVET'
        THEN o.value END) AS tvet_admissions
FROM observations o
JOIN regions r ON o.region_id = r.region_id
JOIN education_levels l ON o.level_id = l.level_id
JOIN indicators i ON o.indicator_id = i.indicator_id
JOIN genders g ON o.gender_id = g.gender_id
WHERE o.year = 2025
  AND i.indicator_name = 'admissions'
  AND g.gender_name = 'Total'
  AND l.level_name IN ('Higher', 'TVET')
  AND r.region_name != 'Kazakhstan'
GROUP BY r.region_name
HAVING COUNT(DISTINCT l.level_name) = 2
ORDER BY higher_admissions DESC;
"""


# Q10 — Historical missing-value audit
# SQL concepts: GROUP BY, SUM CASE, HAVING

sql_queries["Q10_Data_Quality"] = """
SELECT
    l.level_name AS education_level,
    o.value_status,
    COUNT(*) AS records
FROM observations o
JOIN education_levels l ON o.level_id = l.level_id
GROUP BY l.level_name, o.value_status
HAVING COUNT(*) > 0
ORDER BY l.level_name, records DESC;
"""

print("SQL queries defined:", len(sql_queries))
```

```python
# STEP 23B — MATCH GEOJSON TO EDUCATION DATA

import pandas as pd
import re
import unicodedata
import json

region_codes = {
    "KZ10": "Abay",
    "KZ11": "Akmola",
    "KZ15": "Aktobe",
    "KZ19": "Almaty Region",
    "KZ23": "Atyrau",
    "KZ27": "West Kazakhstan",
    "KZ31": "Zhambyl",
    "KZ33": "Zhetysu",
    "KZ35": "Karaganda",
    "KZ39": "Kostanay",
    "KZ43": "Kyzylorda",
    "KZ47": "Mangystau",
    "KZ55": "Pavlodar",
    "KZ59": "North Kazakhstan",
    "KZ61": "Turkestan",
    "KZ62": "Ulytau",
    "KZ63": "East Kazakhstan",
    "KZ71": "Astana",
    "KZ75": "Almaty City",
    "KZ79": "Shymkent"
}

def normalize_name(value):
    if value is None:
        return ""
    text = unicodedata.normalize("NFKC", str(value))
    text = re.sub(r"\s+", " ", text).strip().casefold()
    return text

name_aliases = {
    normalize_name(v): v for v in region_codes.values()
}

name_aliases.update({
    "abai": "Abay",
    "abai region": "Abay",
    "jambyl": "Zhambyl",
    "jambyl region": "Zhambyl",
    "jetisu": "Zhetysu",
    "jetisu region": "Zhetysu",
    "almaty": "Almaty Region",
    "almaty (city)": "Almaty City",
    "astana (city)": "Astana",
    "shymkent (city)": "Shymkent"
})

expected_regions = set(
    analysis_df.loc[
        analysis_df["region"] != "Kazakhstan",
        "region"
    ].unique()
)

mapping_records = []

for feature in features:
    props = feature.get("properties", {})

    # Support different GeoJSON property naming styles
    code = str(props.get("ADM1_PCODE") or "").strip()
    name = str(props.get("ADM1_EN") or "").strip()

    # Prefer a recognized code
    region = region_codes.get(code)

    # Otherwise use the source name
    if region is None:
        region = name_aliases.get(normalize_name(name))

    feature["properties"]["analysis_region"] = region

    mapping_records.append({
        "code": code,
        "source_name": name,
        "matched_region": region
    })

boundary_matches = pd.DataFrame(mapping_records)

display(boundary_matches)

actual_regions = set(
    boundary_matches["matched_region"].dropna()
)

missing = expected_regions - actual_regions
extra = actual_regions - expected_regions
duplicates = boundary_matches[
    boundary_matches["matched_region"].duplicated(
        keep=False
    ) &
    boundary_matches["matched_region"].notna()
]

print("\nExpected regions:", len(expected_regions))
print("Matched regions:", len(actual_regions))
print("Missing:", sorted(missing))
print("Unexpected:", sorted(extra))
print("Duplicate matches:", len(duplicates))

if (
    len(features) != 20
    or missing
    or extra
    or not duplicates.empty
    or boundary_matches["matched_region"].isna().any()
):
    raise ValueError(
        "Boundary matching incomplete. "
        "Inspect the table before making a map."
    )

with open("kazakhstan_regions_2025.geojson", "w") as f:
    json.dump(kazakhstan_geojson, f, ensure_ascii=False)

print("\nSUCCESS: All 20 regions matched.")
```

```python
# STEP 23C — INTERACTIVE FOLIUM MAP

import folium
from folium.features import GeoJsonTooltip
from branca.colormap import linear
from IPython.display import display

def create_education_map(
    year=2025,
    education_level="Higher",
    indicator="enrollment",
    gender="Total"
):
    data = analysis_df[
        (analysis_df["year"] == year) &
        (analysis_df["region"] != "Kazakhstan") &
        (analysis_df["education_level"] == education_level) &
        (analysis_df["indicator"] == indicator) &
        (analysis_df["gender"] == gender) &
        (analysis_df["school_type"] != "Evening")
    ].copy()

    if data.empty:
        raise ValueError("No observations for these filters.")

    if data["region"].duplicated().any():
        raise ValueError("Duplicate regional values detected.")

    values = data.set_index("region")[
        "analysis_value"
    ].to_dict()

    observed = pd.Series(values).dropna()

    if observed.empty:
        raise ValueError("No numeric data to map.")

    vmin = observed.min()
    vmax = observed.max()

    if vmin == vmax:
        vmax = vmin + 1

    colormap = linear.YlOrRd_09.scale(vmin, vmax)
    colormap.caption = (
        f"{education_level} — {indicator} "
        f"({gender}), {year}"
    )

    # Separate map object for each selection
    m = folium.Map(
    location=[48.0, 67.0],
    zoom_start=5,
    tiles="CartoDB positron"
)

    # Copy features to avoid changing original geometry
    geo_data = json.loads(
        json.dumps(kazakhstan_geojson)
    )

    for feature in geo_data["features"]:
        region = feature["properties"]["analysis_region"]
        value = values.get(region, np.nan)

        feature["properties"]["mapped_value"] = (
            "Not available"
            if pd.isna(value)
            else f"{value:,.2f}"
        )

    def style_function(feature):
        region = feature["properties"]["analysis_region"]
        value = values.get(region, np.nan)

        return {
            "fillColor": (
                "#D9D9D9"
                if pd.isna(value)
                else colormap(value)
            ),
            "color": "#555555",
            "weight": 1,
            "fillOpacity": 0.75
        }

    folium.GeoJson(
        geo_data,
        name="Regional statistics",
        style_function=style_function,
        highlight_function=lambda feature: {
            "weight": 3,
            "color": "#222222",
            "fillOpacity": 0.9
        },
        tooltip=GeoJsonTooltip(
            fields=[
                "analysis_region",
                "mapped_value"
            ],
            aliases=[
                "Region:",
                "Indicator value:"
            ],
            localize=True
        )
    ).add_to(m)

    colormap.add_to(m)

    folium.LayerControl().add_to(m)

    return m


education_map = create_education_map(
    year=2025,
    education_level="Higher",
    indicator="enrollment"
)

display(education_map)

education_map.save(
    "kazakhstan_education_map_2025.html"
)

print("Map saved successfully.")
```

```python
# STEP 25: PROJECT VALIDATION AND AUTOMATED TESTS(just for assurance)
from pathlib import Path
import pandas as pd
import sqlite3
import json
import py_compile
import importlib.util
import subprocess
import sys

PROJECT = Path("kazakhstan_education_project").resolve()
DATA = PROJECT / "data"

checks = []

def record_check(name, condition, details=""):
    checks.append({
        "test": name,
        "result": "PASS" if condition else "FAIL",
        "details": str(details)
    })

# 1. Python module syntax
for name in ["app.py", "data_utils.py"]:
    path = PROJECT / name
    try:
        py_compile.compile(str(path), doraise=True)
        record_check(f"{name} syntax", True)
    except Exception as e:
        record_check(f"{name} syntax", False, e)

# 2. Required files
required_files = [
    "education_clean.csv",
    "national_timeseries.csv",
    "forecast_results.csv",
    "forecast_metrics.csv",
    "female_share_2025.csv",
    "kazakhstan_regions.geojson",
    "education.db"
]

for name in required_files:
    record_check(
        f"File: {name}",
        (DATA / name).exists()
    )

# 3. Education data
education = pd.read_csv(DATA / "education_clean.csv")

record_check(
    "Education observations",
    len(education) == 9198,
    f"{len(education)} records"
)

record_check(
    "20 regional units",
    education.loc[
        education["region"] != "Kazakhstan",
        "region"
    ].nunique() == 20
)

# 4. Check national higher enrollment
subset = education.loc[
    (education["year"] == 2025)
    & (education["region"] == "Kazakhstan")
    & (education["education_level"] == "Higher")
    & (education["indicator"] == "enrollment")
    & (education["gender"] == "Total")
]

record_check(
    "2025 higher enrollment",
    len(subset) == 1
    and float(subset["analysis_value"].iloc[0]) == 678114
)

# 5. Geographic data
with open(
    DATA / "kazakhstan_regions.geojson",
    encoding="utf-8"
) as f:
    geojson = json.load(f)

map_regions = [
    feature["properties"].get("analysis_region")
    for feature in geojson["features"]
]

data_regions = set(
    education.loc[
        education["region"] != "Kazakhstan",
        "region"
    ].unique()
)

record_check(
    "20 unique map boundaries",
    len(map_regions) == 20
    and len(set(map_regions)) == 20
)

record_check(
    "Map and statistics match",
    set(map_regions) == data_regions
)

# 6. Database consistency
with sqlite3.connect(DATA / "education.db") as db:

    count = db.execute(
        "SELECT COUNT(*) FROM observations"
    ).fetchone()[0]

    fk_errors = db.execute(
        "PRAGMA foreign_key_check"
    ).fetchall()

    integrity = db.execute(
        "PRAGMA integrity_check"
    ).fetchone()[0]

    tables = pd.read_sql_query("""
        SELECT name
        FROM sqlite_master
        WHERE type = 'table'
          AND name NOT LIKE 'sqlite_%'
    """, db)

record_check(
    "Database records",
    count == len(education),
    count
)

record_check(
    "Seven SQLite tables",
    len(tables) == 7,
    len(tables)
)

record_check(
    "Foreign key integrity",
    len(fk_errors) == 0,
    len(fk_errors)
)

record_check(
    "SQLite integrity",
    integrity == "ok",
    integrity
)

# 7. Gender results
female = pd.read_csv(DATA / "female_share_2025.csv")

record_check(
    "Female-share regional coverage",
    len(female) == 20
    and female["region"].nunique() == 20
)

record_check(
    "Female percentages valid",
    female["female_share_pct"].between(0, 100).all()
)

# 8. Forecast data availability
forecast = pd.read_csv(DATA / "forecast_results.csv")
metrics = pd.read_csv(DATA / "forecast_metrics.csv")

record_check(
    "Forecast results available",
    len(forecast) > 0,
    f"{len(forecast)} rows"
)

record_check(
    "Forecast metrics available",
    len(metrics) > 0,
    f"{len(metrics)} rows"
)

# 9. Dashboard has four required sections
app_text = (PROJECT / "app.py").read_text(
    encoding="utf-8"
)

for section in [
    "Overview",
    "Gender & Geography",
    "Trends & Forecasting",
    "Database Explorer"
]:
    record_check(
        f"Dashboard section: {section}",
        section in app_text
    )

# Save results
tests_df = pd.DataFrame(checks)

tests_df.to_csv(
    PROJECT / "validation_results.csv",
    index=False
)

print("PROJECT VALIDATION RESULTS")
display(tests_df)

passed = (tests_df["result"] == "PASS").sum()
total = len(tests_df)

print(f"\nPassed: {passed}/{total}")

print("\nFORECAST RESULT COLUMNS:")
print(forecast.columns.tolist())

print("\nFORECAST METRICS COLUMNS:")
print(metrics.columns.tolist())

print("\nFORECAST RESULTS PREVIEW:")
display(forecast.head())

if passed != total:
    print("\nWARNING: Review failed tests before submission.")
else:
    print("\nAll automated validation checks passed.")
```

```python
# STEP 27 — CREATE AND DOWNLOAD FINAL PROJECT ZIP

from pathlib import Path
from google.colab import files
import zipfile

PROJECT = Path("/content/kazakhstan_education_project")
ZIP_PATH = Path("/content/Kazakhstan_Education_Final_Project.zip")

# 1. Check that essential project files exist
required = [
    "app.py",
    "data_utils.py",
    "requirements.txt",
    "README.md",
    "data/education_clean.csv",
    "data/national_timeseries.csv",
    "data/forecast_results.csv",
    "data/forecast_metrics.csv",
    "data/female_share_2025.csv",
    "data/kazakhstan_regions.geojson",
    "data/education.db"
]

missing = [
    name for name in required
    if not (PROJECT / name).is_file()
]

if missing:
    raise FileNotFoundError(
        f"Missing project files: {missing}"
    )

# 2. Create ZIP
with zipfile.ZipFile(
    ZIP_PATH, "w", zipfile.ZIP_DEFLATED
) as archive:

    for path in PROJECT.rglob("*"):

        if not path.is_file():
            continue

        # Exclude temporary files
        if (
            "__pycache__" in path.parts
            or path.suffix == ".pyc"
            or path.name == "streamlit.log"
        ):
            continue

        archive.write(
            path,
            arcname=path.relative_to(PROJECT.parent)
        )

# 3. Verify ZIP integrity
with zipfile.ZipFile(ZIP_PATH) as archive:
    assert archive.testzip() is None
    included = archive.namelist()

print("FINAL PROJECT ZIP CREATED SUCCESSFULLY")
print("Filename:", ZIP_PATH.name)
print("Number of files:", len(included))
print(
    "Size:",
    round(ZIP_PATH.stat().st_size / 1024**2, 2),
    "MB"
)

print("\nFiles included:")
for name in included:
    print(name)

# 4. Download ZIP
files.download(str(ZIP_PATH))
```
